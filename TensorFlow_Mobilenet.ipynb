```python
# Steps to take before feeding the image to the model
# Creating a model
# Training a model
# Test or validate 
# Predict 
```

```python
# Pre-processing the image
#second method to load image
from tensorflow.keras.preprocessing import image
img = image.load_img(filename, target_size = (224, 224))
```

```python
plt.imshow(img)
```

```python
resized_img = image.img_to_array(img)
final_image = np.expand_dims(resized_img, axis = 0) #need fourth dimesnsion 
final_image = tf.keras.applications.mobilenet.preprocess_input(final_image)
```

```python
resized_img.shape
```

```python
final_image.shape #adds an extra dimension to the image for deep learning models
```

```python
prediction = mobile.predict(final_image)
```

```python
print(prediction)
```

```python
from tensorflow.keras.applications import imagenet_utils
```

```python
results = imagenet_utils.decode_predictions(prediction, top = 3)

print(results)
```
